```python
rng = np.random.default_rng(13)

Q = 1.0
R = 2.0
velocity = 1
dt = 1.0

model = stats.gaussian(velocity * dt, Q)
x0 = stats.gaussian(0.0, 20.0)

dog = linear.DogSimulation(
    x0=x0.mean, model_mean=model.mean, vel_var=model.sigma, meas_var=R, rng=rng
)
zs = dog.simulate(num=10).z[:, 0]

priors, xs, ps = bayes.prior_estimate(x0, model, zs, R, show_steps=True)
```

```python
rng = np.random.default_rng(6)

Q = 2.0
R = 4.5
model = stats.gaussian(1.0, Q)

dog = linear.DogSimulation(x0=x0.mean, model_mean=model.mean, vel_var=Q, meas_var=R, rng=rng)
tr = dog.simulate(num=25); xs, zs = tr.x[:, 0], tr.z[:, 0]

priors, xs, ps = bayes.prior_estimate(x0, model, zs, R)

_, ax = plt.subplots(figsize=(8, 4))

plot_common.plot_zs(ax, zs)
plot_common.plot_preds(ax, priors[:, 0], kind="scatter")
plot_kf.plot_kf(ax, xs[:, 0], var=priors[:, 1], band_color="orange")
```

```python
rng = np.random.default_rng(7)
actual_voltage, voltage_std = 16.3, 0.13
zs = [datagen.add_jitter(actual_voltage, std=voltage_std, rng=rng) for _ in range(50)]

Q = 0.05
R = voltage_std**2
model = stats.gaussian(0.0, Q)
priors, xs, ps = bayes.prior_estimate(x0, model, zs, R)
xmeans, pmeans = xs[:, 0], ps[:, 1]

_, axes = plt.subplots(2, 1, figsize=(8, 5), constrained_layout=True)

plot_common.plot_zs(axes[0], zs)
plot_kf.plot_kf(axes[0], xmeans, var=pmeans)
axes[0].set(xlabel="Step", ylabel="Volts", ylim=(15.5, 17))
axes[0].legend()
axes[1].plot(ps[:, 1])
axes[1].set(title="Variance")

print(f"Variance converges to {pmeans[-1]:.3f}")
```

### Nonlinear Systems


```python
rng = np.random.default_rng(16)

Q = 2.0
R = 30.0
pos = stats.gaussian(100.0, 500.0**2)
model = stats.gaussian(1.0, Q)

zs, ps = [], []
for i in range(100):
    pos = stats.gaussian_sum(pos, model)
    z = np.sin(i / 3.0) * 2 + rng.standard_normal() * 1.2
    likelihood = stats.gaussian(z, R)
    pos = stats.gaussian_product(pos, likelihood)
    zs.append(z)
    ps.append(pos.mean)


_, ax = plt.subplots(figsize=(8, 4))

ax.plot(zs, c="r", linestyle="--", label="Measured")
ax.plot(ps, c="#004080", label="Filter")
ax.legend();
```

## Multi Dimension


```python
dt = 0.1
kf_cv = const_vel.constant_velocity_filter_2d(P=500, R=5, Q=0.1, dt=dt)
```

```python
rng = np.random.default_rng(19)
def run(zs, x0=(0.0, 0.0), P=500, R=0, Q=0, dt=1.0, **kwargs):
    kf = const_vel.constant_velocity_filter_2d(R=R, P=P, Q=Q, x=x0, dt=dt)
    xs, cov = [], []
    for z in zs:
        kf.predict()
        kf.update(z)
        xs.append(kf.x)
        cov.append(kf.P)

    return np.array(xs), np.array(cov)


R = 10
Q = 0.01

tr = datagen.gen_jittered_vel(x0=0, vel=1, num=50, meas_var=R, vel_var=Q, rng=rng); track, zs = tr.x[:, 0], tr.z[:, 0]

_, ax = plt.subplots(figsize=(8, 4))

P = np.diag([500.0, 49.0])
Ms, Ps = run(zs, count=50, R=10, Q=0.01, P=P)
plot_kf.plot_kf_with_cov(ax, Ms[:, 0], Ps, track, zs)
```

```python
rng = np.random.default_rng(20)
dt = 1.0
R = 10
Q = 0.01
x = np.array([[10.0, 4.5]]).T
P = np.diag([500, 49])
F = np.array([[1, dt], [0, 1]])
H = np.array([[1.0, 0.0]])
R = np.array([[R]])
# Q = discretizer.Q_discrete_white_noise(dim=2, dt=dt, var=Q)

tr = datagen.gen_jittered_vel(x0=0, vel=1, num=50, meas_var=R[0, 0], vel_var=Q, rng=rng); track, zs = tr.x[:, 0], tr.z[:, 0]
Ms, Ps = run(zs, count=50, R=R, Q=Q, P=P)

_, ax = plt.subplots(figsize=(8, 4))
plot_kf.plot_kf_with_cov(ax, Ms[:, 0], Ps, track, zs)
```

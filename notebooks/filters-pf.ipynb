# Particle Filter

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
from plots import plot_pf
from simulators import robot
```

## Localization


```python
_, ax = plt.subplots(figsize=(8, 4))

robot.robot_pf(ax, N=5000, rng=np.random.default_rng(2))
ax.set(xlim=(0, 20), ylim=(0, 20))
```

```python
_, ax = plt.subplots(figsize=(8, 4))

robot.robot_pf(
    ax, N=5000, iters=8, show_particles=True, rng=np.random.default_rng(2)
)
ax.set(xlim=(0, 8), ylim=(0, 8))
```

```python
_, ax = plt.subplots(figsize=(8, 4))

robot.robot_pf(
    ax,
    N=5000,
    show_particles=True,
    initial_x=(1, 1, np.pi / 4),
    rng=np.random.default_rng(6),
)
ax.set(xlim=(0, 20), ylim=(0, 20))
```

## Resampling


```python
a = [0.1, 0.2, 0.3, 0.4, 0.2, 0.3, 0.1]

plot_pf.show_resample_multinomial(a, figsize=(6, 1.5), rng=np.random.default_rng(8))

# plt.savefig("../images/resample_multinomial.png")
```

```python
rng = np.random.default_rng(9)

plot_pf.show_resample_residual(a, figsize=(6, 1.5), rng=rng)

# plt.savefig("../images/resample_residual.png")
```

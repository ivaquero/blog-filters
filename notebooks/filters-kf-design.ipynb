## A Demo KF


```python
R_std = 0.025
Q_std = 1
dt = 1
dim_x = 4
dim_z = 2


def tracker1():
    tracker = kalman.KalmanFilter(dim_x=dim_x, dim_z=dim_z)
    dt = 1.0  # time step
    tracker.F = np.array([[1, dt, 0, 0], [0, 1, 0, 0], [0, 0, 1, dt], [0, 0, 0, 1]])
    tracker.u = 0.0
    tracker.H = np.array([[1 / 0.3048, 0, 0, 0], [0, 0, 1 / 0.3048, 0]])
    tracker.R = np.eye(dim_z) * R_std**2
    q = noise.white_noise_discrete(dim=2, dt=1, var=Q_std**2)
    tracker.Q = linalg.block_diag(q, q)
    tracker.P = np.eye(dim_x) * 50.0
    tracker.x = np.array([[0, 0, 0, 0]]).T
    return tracker
```

```python
rng = np.random.default_rng(6)

N = 50
sensor = linear.PosSensor((0, 0), (2, 0.2), R=R_std, rng=rng)

zs = np.array([sensor.read() for _ in range(N)])

# run filter
robot_tracker = tracker1()
mu, cov, _, _ = robot_tracker.batch_filter(zs)

_, ax = plt.subplots(figsize=(8, 4))

for x, P in zip(mu, cov):
    # covariance of x and y
    cov_xy = np.array([[P[0, 0], P[2, 0]], [P[0, 2], P[2, 2]]])
    mean = (x[0, 0], x[2, 0])
    plot_common.plot_cov_ellipse(
        ax, mean, cov=cov_xy, facecolor="g", stds=[3], alpha=0.5
    )

# plot results
zs *= 0.3048  # convert to meters
plot_common.plot_zs(ax, zs[:, 0], zs[:, 1])
plot_kf.plot_kf(ax, mu[:, 0], mu[:, 2])
```

## Varying Order


```python
def filter_data(kf, zs):
    s = helpers.KFSaver(kf)
    kf.batch_filter(zs, saver=s)
    s.to_array()
    return s
```

### 1st Order


```python
rng = np.random.default_rng(11)

R, Q = 1, 0.03
cvobj = linear.CVObject(Q=Q, R=R, rng=rng)

xs, zs = cvobj.simulate(count=50)
```

```python
P = np.array([[100, 0], [0, 1]])
kf = const_vel.constant_velocity_filter_2d(P=P, R=R, Q=Q, dt=1)
data = filter_data(kf, zs)

_, ax = plt.subplots(figsize=(8, 4))

plot_kf.plot_kf_track(ax, xs, data.x, data.z)
```

## Performance


### State Variable


```python
# `CAObject` takes standard deviations; the filter takes variances, hence the
# squaring below. Passing R=6 straight through leaves the filter overconfident
# (R 36x too small) and the NEES consistency test fails.
R_std, Q_std = 6.0, 0.02
n_runs, n_steps = 100, 80

nees = []
for seed in range(n_runs):
    caobj = linear.CAObject(Q=Q_std, R=R_std, rng=np.random.default_rng(seed))
    xs, zs = caobj.simulate(count=n_steps)

    kf2 = const_acc.KFCA3d(P=100, R=R_std**2, Q=Q_std**2, dt=1)
    est_xs, ps, _, _ = kf2.batch_filter(zs)
    nees.extend(stats.NEES(xs, est_xs, ps))

eps = np.mean(nees)
print(f"mean NEES is: {eps:.3f} (theoretical mean = dim_x = {kf2.dim_x})")

if eps < kf2.dim_x:
    print("passed")
else:
    print("failed")
```
